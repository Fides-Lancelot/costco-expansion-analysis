```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from datetime import datetime

# Cargar datos
df = pd.read_csv(r'..\data\processed\datos_modelo.csv', encoding='utf-8-sig')
df.columns = df.columns.str.replace('ï»¿', '', regex=False)

# Limpiar y crear variables
df_clean = df.dropna(subset=['ingreso_total']).copy()
df_clean['log_ingreso'] = np.log(df_clean['ingreso_total'] + 1)
df_clean['est_socio_num'] = pd.to_numeric(df_clean['est_socio'], errors='coerce')
df_clean['tam_loc_num'] = pd.to_numeric(df_clean['tam_loc'], errors='coerce')
df_clean['entidad'] = (df_clean['ubica_geo'] // 1000).astype(int)

# Nuevo León es la categoría base
# Las dummies comparan Jalisco y Querétaro contra Nuevo León
df_clean['es_jalisco'] = (df_clean['entidad'] == 14).astype(int)
df_clean['es_queretaro'] = (df_clean['entidad'] == 22).astype(int)

df_model = df_clean.dropna(subset=['log_ingreso', 'est_socio_num', 'tam_loc_num']).copy()

# Especificar y estimar el modelo OLS
y = df_model['log_ingreso']
X = df_model[['est_socio_num', 'tam_loc_num', 'es_jalisco', 'es_queretaro']]
X = sm.add_constant(X)
modelo = sm.OLS(y, X).fit()

# Reporte profesional
print("\n" + "=" * 80)
print("ANÁLISIS ECONOMÉTRICO: DEMANDA POTENCIAL DE COSTCO EN MÉXICO")
print("=" * 80)
print(f"\nINFORMACIÓN DEL MODELO")
print("-" * 80)
print(f"Fecha de análisis:        {datetime.now().strftime('%d/%m/%Y %H:%M')}")
print(f"Número de observaciones:  {len(df_model):,} hogares")
print("Ciudades analizadas:      Monterrey, Guadalajara y Querétaro")
print("Fuente de datos:          INEGI ENIGH 2022")

print("\nDISTRIBUCIÓN POR ENTIDAD")
print("-" * 80)
print(df_model['entidad'].value_counts().sort_index())

print("\nBONDAD DEL MODELO")
print("-" * 80)
print(f"R-cuadrado (R²):          {modelo.rsquared:.4f} ({modelo.rsquared * 100:.2f}%)")
print(f"F-statistic:              {modelo.fvalue:.2f}")
print(f"Prob (F-statistic):       {modelo.f_pvalue:.2e}")

print("\nCOEFICIENTES E INTERVALOS DE CONFIANZA")
print("-" * 80)
print(modelo.conf_int().rename(columns={0: 'IC_95_inferior', 1: 'IC_95_superior'}))

print("\nRESULTADOS COMPLETOS OLS")
print("-" * 80)
print(modelo.summary())

print("\nLIMITACIONES")
print("-" * 80)
print(f"- El modelo explica {modelo.rsquared * 100:.2f}% de la variación observada.")
print("- No incluye PIB regional, competencia, densidad comercial ni costos de terreno.")
print("- Las diferencias estatales deben interpretarse respecto a Nuevo León.")

print("\nRECOMENDACIÓN PRELIMINAR")
print("-" * 80)
print("El estrato socioeconómico es la variable más relevante del modelo.")
print("Se requieren variables macroeconómicas y comerciales adicionales")
print("antes de elegir definitivamente entre las tres ciudades.")
print("\n" + "=" * 80)
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from sklearn.preprocessing import StandardScaler

# Cargar datos
df = pd.read_csv(r'..\data\processed\datos_modelo.csv', encoding='utf-8-sig')
df.columns = df.columns.str.replace('ï»¿', '', regex=False)

# Eliminar valores faltantes en ingreso
df_clean = df.dropna(subset=['ingreso_total']).copy()

print("Filas después de limpiar:", len(df_clean))

# CREAR VARIABLE DEPENDIENTE: log(Ingreso)
df_clean['log_ingreso'] = np.log(df_clean['ingreso_total'] + 1)

# CREAR VARIABLES INDEPENDIENTES
# 1. Estrato socioeconómico (proxy de clase social/ingresos)
df_clean['est_socio_num'] = pd.to_numeric(df_clean['est_socio'], errors='coerce')

# 2. Tamaño de localidad (proxy de urbanización)
df_clean['tam_loc_num'] = pd.to_numeric(df_clean['tam_loc'], errors='coerce')

# 3. Dummies por estado
df_clean['es_nuevo_leon'] = (df_clean['ubica_geo'].astype(str).str[:2] == '19').astype(int)
df_clean['es_jalisco'] = (df_clean['ubica_geo'].astype(str).str[:2] == '14').astype(int)
df_clean['es_queretaro'] = (df_clean['ubica_geo'].astype(str).str[:2] == '22').astype(int)

# Eliminar NaN en variables independientes
df_model = df_clean.dropna(subset=['log_ingreso', 'est_socio_num', 'tam_loc_num'])

print("Filas para modelo:", len(df_model))

# ESPECIFICAR EL MODELO
# Variable dependiente
y = df_model['log_ingreso']

# Variables independientes
X = df_model[['est_socio_num', 'tam_loc_num', 'es_nuevo_leon', 'es_jalisco', 'es_queretaro']]

# Añadir constante
X = sm.add_constant(X)

# CORRER LA REGRESIÓN
modelo = sm.OLS(y, X).fit()

# VER RESULTADOS
print("\n" + "=" * 70)
print("MODELO ECONOMÉTRICO: DEMANDA DE COSTCO")
print("=" * 70)
print(modelo.summary())

# GUARDAR RESULTADOS
with open(r'..\resultados\modelo_resultado.txt', 'w', encoding='utf-8') as archivo:
    archivo.write(modelo.summary().as_text())
print("\nResultados guardados en modelo_resultado.txt")
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm

# Cargar datos
df = pd.read_csv(r'..\data\processed\datos_modelo.csv', encoding='latin-1')

# Limpiar
df_clean = df.dropna(subset=['ingreso_total'])

# VARIABLE DEPENDIENTE
df_clean['log_ingreso'] = np.log(df_clean['ingreso_total'] + 1)

# VARIABLES INDEPENDIENTES
df_clean['est_socio_num'] = pd.to_numeric(df_clean['est_socio'], errors='coerce')
df_clean['tam_loc_num'] = pd.to_numeric(df_clean['tam_loc'], errors='coerce')

# EXTRAER ENTIDAD (primeros 2 dígitos de ubica_geo)
df_clean['entidad'] = (df_clean['ubica_geo'] // 1000).astype(int)

# Dummies (sin Nuevo León como base para evitar multicolinealidad)
df_clean['es_jalisco'] = (df_clean['entidad'] == 14).astype(int)
df_clean['es_queretaro'] = (df_clean['entidad'] == 22).astype(int)

# Limpiar NaN
df_model = df_clean.dropna(subset=['log_ingreso', 'est_socio_num', 'tam_loc_num'])

print("Filas para modelo:", len(df_model))
print("\nDistribución por entidad:")
print(df_model['entidad'].value_counts().sort_index())

# MODELO ECONOMÉTRICO CORREGIDO
y = df_model['log_ingreso']
X = df_model[['est_socio_num', 'tam_loc_num', 'es_jalisco', 'es_queretaro']]
X = sm.add_constant(X)

modelo = sm.OLS(y, X).fit()

print("\n" + "="*70)
print("MODELO ECONOMÉTRICO: DEMANDA DE COSTCO")
print("="*70)
print(modelo.summary())
```

```python
# Ver los coeficientes completos
print("COEFICIENTES DEL MODELO:")
print(modelo.params)
print("\nP-VALUES:")
print(modelo.pvalues)
print("\nINTERVALOS DE CONFIANZA (95%):")
print(modelo.conf_int())
```

```python
"""
================================================================================
ANÁLISIS ECONOMÉTRICO: DEMANDA DE COSTCO EN MÉXICO
Fase 1: Exploración de Datos y Modelo Econométrico
================================================================================
Autor: Hugo Cuellar
Fecha: 24/08/2026
Fuente: INEGI ENIGH 2022
Ciudades: Monterrey (NL), Guadalajara (Jalisco), Querétaro
================================================================================
"""

import pandas as pd
import numpy as np
import statsmodels.api as sm
from datetime import datetime
import warnings

warnings.filterwarnings('ignore')

print("\n" + "=" * 80)
print("INICIANDO ANÁLISIS ECONOMÉTRICO DE COSTCO")
print("=" * 80)

# PASO 1: CARGAR DATOS
print("\nPASO 1: Cargando datos...")

df_hogares = pd.read_csv(
    r'..\data\processed\enigh_monterrey_guadalajara_queretaro.csv',
    encoding='utf-8-sig'
)
df_hogares.columns = df_hogares.columns.str.replace('ï»¿', '', regex=False)

df_ingresos = pd.read_csv(
    r'..\data\processed\ingresos_por_hogar.csv'
)

print(f"Hogares cargados: {len(df_hogares):,} registros, {len(df_hogares.columns)} columnas")
print(f"Ingresos cargados: {len(df_ingresos):,} registros, {len(df_ingresos.columns)} columnas")

# PASO 2: UNIR TABLAS
print("\nPASO 2: Uniendo tablas de hogares e ingresos...")

df_merged = df_hogares.merge(
    df_ingresos[['folioviv', 'foliohog', 'ingreso_total']],
    on=['folioviv', 'foliohog'],
    how='left'
)

print(f"Tabla unida: {len(df_merged):,} registros")
print(f"Valores faltantes en ingresos: {df_merged['ingreso_total'].isnull().sum()}")

# PASO 3: LIMPIEZA DE DATOS
print("\nPASO 3: Limpiando datos...")
df_clean = df_merged.dropna(subset=['ingreso_total']).copy()
print(f"Datos limpios: {len(df_clean):,} registros")

# PASO 4: CREAR VARIABLES PARA EL MODELO
print("\nPASO 4: Creando variables para el modelo...")
df_clean['log_ingreso'] = np.log(df_clean['ingreso_total'] + 1)
df_clean['est_socio_num'] = pd.to_numeric(df_clean['est_socio'], errors='coerce')
df_clean['tam_loc_num'] = pd.to_numeric(df_clean['tam_loc'], errors='coerce')
df_clean['entidad'] = (df_clean['ubica_geo'] // 1000).astype(int)

# Nuevo León es la categoría base.
df_clean['es_jalisco'] = (df_clean['entidad'] == 14).astype(int)
df_clean['es_queretaro'] = (df_clean['entidad'] == 22).astype(int)

df_model = df_clean.dropna(
    subset=['log_ingreso', 'est_socio_num', 'tam_loc_num']
).copy()
print("Variables creadas")
print(f"Dataset final para modelo: {len(df_model):,} registros")

# PASO 5: CONSTRUIR Y EJECUTAR MODELO ECONOMÉTRICO
print("\nPASO 5: Construyendo modelo econométrico...")
y = df_model['log_ingreso']
X = df_model[['est_socio_num', 'tam_loc_num', 'es_jalisco', 'es_queretaro']]
X = sm.add_constant(X)
modelo = sm.OLS(y, X).fit()
print("Modelo construido exitosamente")

# PASO 6: REPORTE PROFESIONAL
print("\n" + "=" * 80)
print("ANÁLISIS ECONOMÉTRICO: DEMANDA POTENCIAL DE COSTCO EN MÉXICO")
print("=" * 80)

print("\nINFORMACIÓN DEL MODELO")
print("-" * 80)
print(f"Fecha de análisis:        {datetime.now().strftime('%d/%m/%Y %H:%M')}")
print(f"Número de observaciones:  {len(df_model):,} hogares")
print("Ciudades analizadas:      Monterrey (NL), Guadalajara (Jalisco), Querétaro")
print("Fuente de datos:          INEGI ENIGH 2022")

print("\nBONDAD DEL MODELO")
print("-" * 80)
print(f"R-cuadrado (R²):          {modelo.rsquared:.4f} ({modelo.rsquared * 100:.2f}%)")
print(f"F-statistic:              {modelo.fvalue:.2f}")
print(f"Prob (F-statistic):       {modelo.f_pvalue:.2e}")
print("El modelo conjunto es estadísticamente significativo si p < 0.05.")

print("\nDISTRIBUCIÓN POR ENTIDAD")
print("-" * 80)
print(df_model['entidad'].value_counts().sort_index())

print("\nESTADÍSTICAS DE INGRESOS")
print("-" * 80)
print(df_model['ingreso_total'].describe())

print("\nCOEFICIENTES E INTERPRETACIÓN")
print("-" * 80)

const = modelo.params['const']
print(f"\n1. CONSTANTE: {const:.4f}")
print(f"   P-value: {modelo.pvalues['const']:.2e}")

est = modelo.params['est_socio_num']
est_p = modelo.pvalues['est_socio_num']
print(f"\n2. ESTRATO SOCIOECONÓMICO: {est:.4f}")
print(f"   Cada nivel de estrato se asocia con {est * 100:.2f}% de cambio en el ingreso logarítmico.")
print(f"   P-value: {est_p:.2e}")

tam = modelo.params['tam_loc_num']
tam_p = modelo.pvalues['tam_loc_num']
print(f"\n3. TAMAÑO DE LOCALIDAD: {tam:.4f}")
print(f"   P-value: {tam_p:.4f}")

jal = modelo.params['es_jalisco']
jal_p = modelo.pvalues['es_jalisco']
print(f"\n4. EFECTO JALISCO (Guadalajara): {jal:.4f}")
print("   Comparación respecto a Nuevo León, la categoría base.")
print(f"   P-value: {jal_p:.4f}")

quet = modelo.params['es_queretaro']
quet_p = modelo.pvalues['es_queretaro']
print(f"\n5. EFECTO QUERÉTARO: {quet:.4f}")
print("   Comparación respecto a Nuevo León, la categoría base.")
print(f"   P-value: {quet_p:.4f}")

print("\nLIMITACIONES DEL MODELO")
print("-" * 80)
print(f"- R² de {modelo.rsquared:.2%}: faltan variables explicativas importantes.")
print("- No incluye PIB regional, crecimiento, competencia ni densidad comercial.")
print("- Las diferencias estatales se interpretan respecto a Nuevo León.")

print("\nRECOMENDACIÓN PRELIMINAR")
print("-" * 80)
print("El estrato socioeconómico es la variable más relevante del modelo.")
print("Se requieren variables macroeconómicas y comerciales adicionales")
print("antes de elegir definitivamente entre las tres ciudades.")

print("\n" + "=" * 80)
print("ANÁLISIS FASE 1 COMPLETADO")
print("=" * 80)

# GUARDAR RESULTADOS
print("\nGuardando resultados...")
df_model.to_csv(
    r'..\data\processed\datos_modelo_final.csv',
    index=False
)

with open(
    r'..\resultados\modelo_resumen.txt',
    'w',
    encoding='utf-8'
) as archivo:
    archivo.write("ANÁLISIS ECONOMÉTRICO: DEMANDA DE COSTCO\n")
    archivo.write("=" * 80 + "\n\n")
    archivo.write(modelo.summary().as_text())

print("Archivos guardados en resultados/ y data/processed/")
```
