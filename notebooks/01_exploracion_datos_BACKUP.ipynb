```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from datetime import datetime

# Cargar datos
df = pd.read_csv(r'C:\Users\hugod\costco-expansion-analysis\data\processed\datos_modelo.csv', encoding='utf-8-sig')
df.columns = df.columns.str.replace('ï»¿', '', regex=False)

# Limpiar y crear variables
df_clean = df.dropna(subset=['ingreso_total']).copy()
df_clean['log_ingreso'] = np.log(df_clean['ingreso_total'] + 1)
df_clean['est_socio_num'] = pd.to_numeric(df_clean['est_socio'], errors='coerce')
df_clean['tam_loc_num'] = pd.to_numeric(df_clean['tam_loc'], errors='coerce')
df_clean['entidad'] = (df_clean['ubica_geo'] // 1000).astype(int)

# Nuevo León es la categoría base
# Las dummies comparan Jalisco y Querétaro contra Nuevo León
df_clean['es_jalisco'] = (df_clean['entidad'] == 14).astype(int)
df_clean['es_queretaro'] = (df_clean['entidad'] == 22).astype(int)

df_model = df_clean.dropna(subset=['log_ingreso', 'est_socio_num', 'tam_loc_num']).copy()

# Especificar y estimar el modelo OLS
y = df_model['log_ingreso']
X = df_model[['est_socio_num', 'tam_loc_num', 'es_jalisco', 'es_queretaro']]
X = sm.add_constant(X)
modelo = sm.OLS(y, X).fit()

# Reporte profesional
print("\n" + "=" * 80)
print("ANÁLISIS ECONOMÉTRICO: DEMANDA POTENCIAL DE COSTCO EN MÉXICO")
print("=" * 80)
print(f"\nINFORMACIÓN DEL MODELO")
print("-" * 80)
print(f"Fecha de análisis:        {datetime.now().strftime('%d/%m/%Y %H:%M')}")
print(f"Número de observaciones:  {len(df_model):,} hogares")
print("Ciudades analizadas:      Monterrey, Guadalajara y Querétaro")
print("Fuente de datos:          INEGI ENIGH 2022")

print("\nDISTRIBUCIÓN POR ENTIDAD")
print("-" * 80)
print(df_model['entidad'].value_counts().sort_index())

print("\nBONDAD DEL MODELO")
print("-" * 80)
print(f"R-cuadrado (R²):          {modelo.rsquared:.4f} ({modelo.rsquared * 100:.2f}%)")
print(f"F-statistic:              {modelo.fvalue:.2f}")
print(f"Prob (F-statistic):       {modelo.f_pvalue:.2e}")

print("\nCOEFICIENTES E INTERVALOS DE CONFIANZA")
print("-" * 80)
print(modelo.conf_int().rename(columns={0: 'IC_95_inferior', 1: 'IC_95_superior'}))

print("\nRESULTADOS COMPLETOS OLS")
print("-" * 80)
print(modelo.summary())

print("\nLIMITACIONES")
print("-" * 80)
print(f"- El modelo explica {modelo.rsquared * 100:.2f}% de la variación observada.")
print("- No incluye PIB regional, competencia, densidad comercial ni costos de terreno.")
print("- Las diferencias estatales deben interpretarse respecto a Nuevo León.")

print("\nRECOMENDACIÓN PRELIMINAR")
print("-" * 80)
print("El estrato socioeconómico es la variable más relevante del modelo.")
print("Se requieren variables macroeconómicas y comerciales adicionales")
print("antes de elegir definitivamente entre las tres ciudades.")
print("\n" + "=" * 80)
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from sklearn.preprocessing import StandardScaler

# Cargar datos
df = pd.read_csv(r'C:\Users\hugod\costco-expansion-analysis\data\processed\datos_modelo.csv', encoding='utf-8-sig')
df.columns = df.columns.str.replace('ï»¿', '', regex=False)

# Eliminar valores faltantes en ingreso
df_clean = df.dropna(subset=['ingreso_total']).copy()

print("Filas después de limpiar:", len(df_clean))

# CREAR VARIABLE DEPENDIENTE: log(Ingreso)
df_clean['log_ingreso'] = np.log(df_clean['ingreso_total'] + 1)

# CREAR VARIABLES INDEPENDIENTES
# 1. Estrato socioeconómico (proxy de clase social/ingresos)
df_clean['est_socio_num'] = pd.to_numeric(df_clean['est_socio'], errors='coerce')

# 2. Tamaño de localidad (proxy de urbanización)
df_clean['tam_loc_num'] = pd.to_numeric(df_clean['tam_loc'], errors='coerce')

# 3. Dummies por estado
df_clean['es_nuevo_leon'] = (df_clean['ubica_geo'].astype(str).str[:2] == '19').astype(int)
df_clean['es_jalisco'] = (df_clean['ubica_geo'].astype(str).str[:2] == '14').astype(int)
df_clean['es_queretaro'] = (df_clean['ubica_geo'].astype(str).str[:2] == '22').astype(int)

# Eliminar NaN en variables independientes
df_model = df_clean.dropna(subset=['log_ingreso', 'est_socio_num', 'tam_loc_num'])

print("Filas para modelo:", len(df_model))

# ESPECIFICAR EL MODELO
# Variable dependiente
y = df_model['log_ingreso']

# Variables independientes
X = df_model[['est_socio_num', 'tam_loc_num', 'es_nuevo_leon', 'es_jalisco', 'es_queretaro']]

# Añadir constante
X = sm.add_constant(X)

# CORRER LA REGRESIÓN
modelo = sm.OLS(y, X).fit()

# VER RESULTADOS
print("\n" + "=" * 70)
print("MODELO ECONOMÉTRICO: DEMANDA DE COSTCO")
print("=" * 70)
print(modelo.summary())

# GUARDAR RESULTADOS
with open(r'C:\Users\hugod\costco-expansion-analysis\resultados\modelo_resultado.txt', 'w', encoding='utf-8') as archivo:
    archivo.write(modelo.summary().as_text())
print("\nResultados guardados en modelo_resultado.txt")
```
